```python
# Random and reproducible tensors

import torch
# Set the random seed

RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)   #Here we use two same call becaues it only works for the one block of the code and when you do it againg u have to call it again

random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)    #every time rand is called you have to use

random_tensor_D = torch.rand(3, 4)

print(random_tensor_C == random_tensor_D)
```
